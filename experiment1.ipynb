```python
import numpy as np
import pandas as pd
```

```python
# Create a NumPy array
arr = np.array([78, 92, 65, 84, 71])

# Display the array
print("Array:", arr)

# Calculate mean
print("Mean:", arr.mean())

# Reshape into 2D array
print("Reshaped 2D:")
print(arr.reshape(5, 1))

# Slice elements from index 1 to 3
print("Sliced (index 1 to 3):", arr[1:4])
```

```python
# Create student data
data = {
    "Student": ["Arjun", "Meera", "Vikram", "Neha"],
    "Attendance": [85, 92, 78, 95],
    "Score": [76, 89, 68, 94]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display DataFrame
print("Student Data:")
display(df)
```

```python
print("First two rows:")
display(df.iloc[0:2])
```

```python
print("Score column:")
display(df["Score"])
```

```python
print("Rows where Score > 80:")
display(df[df["Score"] > 80])
```
